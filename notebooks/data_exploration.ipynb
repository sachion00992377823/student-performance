```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 300

data = pd.DataFrame({
    'StudentID': range(1, n + 1),
    'Age': np.random.randint(15, 19, n),
    'Gender': np.random.choice(['Male', 'Female'], n),
    'StudyHoursPerDay': np.round(np.random.uniform(1, 10, n), 1),
    'AttendancePercent': np.round(np.random.uniform(50, 100, n), 1),
    'PreviousScore': np.random.randint(40, 100, n),
    'ExtracurricularActivities': np.random.choice(['Yes', 'No'], n),
    'SleepHours': np.round(np.random.uniform(4, 9, n), 1),
    'FinalResult': np.random.choice(['Pass', 'Fail'], n, p=[0.75, 0.25])
})

data.to_csv('../data/student_data.csv', index=False)
print(f'Dataset created! Shape: {data.shape}')
print(data.head(10))
```

```python

print(f'Total Students: {data.shape[0]}')
print(f'Total Features: {data.shape[1]}')
print('\nMissing Values:')
print(data.isnull().sum())
print('\nData Types:')
print(data.dtypes)
print('\nStatistical Summary:')
data.describe().round(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8')
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Student Performance Analysis', fontsize=16, fontweight='bold')

# Pass vs Fail count
sns.countplot(x='FinalResult', data=data, palette='viridis', ax=axes[0,0])
axes[0,0].set_title('Pass vs Fail Distribution')
axes[0,0].set_xlabel('Result')
axes[0,0].set_ylabel('Count')

# Study hours distribution
axes[0,1].hist(data['StudyHoursPerDay'], bins=20, color='#3498db', edgecolor='white')
axes[0,1].set_title('Study Hours Distribution')
axes[0,1].set_xlabel('Hours per Day')
axes[0,1].set_ylabel('Count')

# Attendance vs Final Result
sns.boxplot(x='FinalResult', y='AttendancePercent', data=data, palette='viridis', ax=axes[1,0])
axes[1,0].set_title('Attendance vs Result')

# Previous Score vs Final Result
sns.boxplot(x='FinalResult', y='PreviousScore', data=data, palette='viridis', ax=axes[1,1])
axes[1,1].set_title('Previous Score vs Result')

plt.tight_layout()
plt.show()
```
