```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split

data = pd.read_csv('../data/student_data.csv')
print('Dataset loaded successfully!')
print(f'Shape: {data.shape}')
data.head()
```

```python

le = LabelEncoder()

data['Gender'] = le.fit_transform(data['Gender'])
data['ExtracurricularActivities'] = le.fit_transform(data['ExtracurricularActivities'])
data['FinalResult'] = le.fit_transform(data['FinalResult'])

print('Encoding done!')
print('\nSample after encoding:')
print(data.head())
print('\nFinalResult values (0=Fail, 1=Pass):')
print(data['FinalResult'].value_counts())
```

```python

X = data.drop(['StudentID', 'FinalResult'], axis=1)
y = data['FinalResult']

print('Features (X):')
print(X.columns.tolist())
print(f'\nTarget (y): FinalResult')
print(f'X Shape: {X.shape}')
print(f'y Shape: {y.shape}')
```

```python

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print('Scaling done!')
print('\nBefore scaling (first 3 rows):')
print(X.head(3).to_string())
print('\nAfter scaling (first 3 rows):')
print(pd.DataFrame(X_scaled, columns=X.columns).head(3).round(4).to_string())
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print('Train/Test Split done!')
print(f'\nTotal Students: {len(X_scaled)}')
print(f'Training Set: {X_train.shape[0]} students (80%)')
print(f'Testing Set: {X_test.shape[0]} students (20%)')

# Save preprocessed data
np.save('../data/X_train.npy', X_train)
np.save('../data/X_test.npy', X_test)
np.save('../data/y_train.npy', y_train)
np.save('../data/y_test.npy', y_test)

print('\nPreprocessed data saved successfully!')
```
